```python
# 결측치를 최빈값으로 채운다.
df['region'] = df['region'].fillna(df['region'].mode()[0])
```

```python
num_cols_na
```

```python
# 숫자형 컬럼의 결측치를 중앙값으로 채운다.
for c in num_cols_na:
    df[c] = df[c].fillna(df[c].median())

df.isnull().sum() # 모두 0이 나오는지 확인
```

### 이상치 처리

```python
# 의료비(charges) 박스 플롯 확인
sns.boxplot(x=df['charges'])
plt.show()
```

```python
# Q1 (1사분위수) 25%
q1 = df['charges'].quantile(.25)

# Q3 (3사분위수) 75%
q3 = df['charges'].quantile(.75)

q1, q3
```

```python
iqr = q3 - q1
iqr
```

```python
# 하한
lower = q1 - 1.5 * iqr
lower
```

```python
# 상한
upper = q3 + 1.5 * iqr
upper
```

```python
# 이상치의 개수
n_outlier = ((df['charges'] < lower) | (df['charges'] > upper)).sum()
n_outlier
```

```python
# 의료비 데이터는 표본 수가 적을수록 뒤쪽 EDA(그룹 비교)에서 신뢰도가 떨어지므로 표본을 유지하면서
# 극단값만 완화하는 capping이 더 적합하다. 
# 고액 의료비를 내는 사람이 흡연자일 수도 있다.  
# 
# .clip() -> 값의 범위를 강제로 잘내내는 함수
#           범위 밖으로 튀어나온 값을 범위 안으로 눌러 넣는다.
#           값이 lower보다 작으면 lower로 바꾼다.
#           값이 upper보다 크면 upper로 바꾼다.
#           그 사이에 있으면 그대로 둔다.   
df['charges'] = df['charges'].clip(lower, upper) 
df['charges']
```

####  왜도 .skew()
- 데이터 분포가 좌우 대칭에서 얼마나, 어느 방향으로 치우쳐 있는지를 나타내는 지표
- 왜도가 0 --> 좌우 대칭(정규분포처럼)
- 왜도 > 0 --> 양의 왜도, 오른쪽으로 긴 꼬리 (대부분의 값은 작은 쪽에 몰려있다. 소수의 큰 값이 꼬리를 늘린다.)
- 왜도 < 0 --> 음의 왜도, 왼쪽으로 긴 꼬리 (대부분의 값은 큰 쪽에 몰려있다. 소수의 작은 값이 꼬리를 늘린다.)

```python
# 의료비의 왜도
df['charges'].skew().round(3)  # 소수 셋째자리까지 반올림
```

### 상관관계 분석

```python
corr = df[['age', 'bmi', 'children', 'charges']].corr(numeric_only=True)

# 히트맵 그래프
sns.heatmap(data=corr, annot=True, fmt='.2f', cmap='coolwarm')
plt.title('수치형 변수 상관관계')
plt.show()
```

```python
np.ones(corr.shape)  # 모두 1로 채운다.
```

```python
# triu --> 위쪽 삼각형의 약자 --> 대각선의 위쪽만 남기고 나머지는 0으로 만든다. 대각선 제외하고 그 위쪽
np.triu(np.ones(corr.shape), k=1)
```

```python
np.triu(np.ones(corr.shape), k=1).astype(bool) # bool형으로 형변환
```

```python
corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool)) # True는 값이 나오고, False는 결측치(NaN)
```

```python
corr_pairs = corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool)).stack() # 2차원표를 1차원 시리즈로 쌓아 편다.
```

```python
# abs() --> 절대값
# sort_values(ascending=False) --> 값을 내림차순 정렬
top_pair = corr_pairs.abs().sort_values(ascending=False).index[0]
print(f'최고 상관 변수 쌍 : {top_pair} r={corr.loc[top_pair]}')
```

```python
top_pair
```

```python
# 산포그래프 - sns.scatterplot()
sns.scatterplot(data=df, x=top_pair[0], y=top_pair[1])
plt.title(f'{top_pair[0]} vs {top_pair[1]}')
```
