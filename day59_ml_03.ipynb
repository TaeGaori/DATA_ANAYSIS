## 로지스틱 회귀
- 0~1 사이의 확률을 출력하고, 이를 임계값으로 잘라서 참(1)/거짓(0)으로 분류한다.

```python
from pathlib import Path
import warnings
warnings.filterwarnings('ignore')

DATA_DIR = Path('ml_실습데이터')
if not DATA_DIR.exists():
    DATA_DIR = Path('ml_실습데이터')

print(f'데이터 폴더 : {DATA_DIR.resolve()}')
```

## 라이브러리 불러오기

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression #로지스틱 회귀 분류 모델
from sklearn.impute import SimpleImputer # 결측치 특정 값으로 채워주는 전처리 클래스
from sklearn.pipeline import make_pipeline  # 여러 전처리 단계와 모델을 하나로 연결해주는 클래스
from sklearn.metrics import confusion_matrix, classification_report # 분류 성능을 평가하는 지표함수들
```

```python
df =pd.read_csv(DATA_DIR / 'telecom_churn.csv') # 통신사 고객 이탈 데이터 불러오기
df.head()
```

```python
# 예측에 사용할 입력 변수(특성/feature) 4개를 선정
features = ['usage_minutes', 'complaints',	'contract_months',	'monthly_fee']

X = df[features]
y= df['churn']

X_train, X_valid, y_train, y_valid = train_test_split(
    X, y, test_size=0.25, stratify=y ,random_state=42
)

X_train.shape, X_valid.shape, y_train.shape, y_valid.shape
```

```python
model = make_pipeline(
    SimpleImputer(strategy='median'),    # 결측치를 중앙값으로 대체 (평균보다 이상치에 덜 민감)
    LogisticRegression(max_iter=1000)   # 반복 횟수를 늘려 실패 경고 방지
)
model.fit(X_train, y_train) # 학습
```

```python
proba = model.predict_proba(X_valid)[:, 1]  # [비이탈확률, 이탈확률] 두 열 반환 --> 이탈(1) 확률 알기

# 임게값 (threshold) 0.5
pred = (proba >= 0.5).astype(int)   # 0.5이상인 것만 추려서 정수형으로 형 변환
pred
```

### 혼동행렬 (confusion_matrix)

[[TN, FP]],

[FN, TP]

- TP : 실제 이탈을 이탈로 정확히 맞췄다
- TN : 실제 비이탈을 비이탈로 정확히 맞췄다
- FP : 실제 비이탈을 이탈로 잘못 예측
- FN : 실제 이탈을 비이탈로 잘못 예측

```python
print(confusion_matrix(y_valid, pred))
```

```python
print(classification_report(y_valid, pred, digits=3))   # 소수 3째자리까지 
# 정밀도, 재현율, 조화평균, 클래스별 표본 수
```

### k-means clustering (k-평균 군집화)
- 중심점을 무작위로 놓는다 -> 가까운 점들을 그 중심에 배정 -> 배정된 점들의 평균으로 중심을 다시 계산
- 위의 과정을 계속해서 반복한다.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_samples  # 군집이 얼마나 잘 분리되어 있나를 -1 ~ 1 값으로 나타낸다.
# 1에 가까울수록 좋다
```

```python
df = pd.read_csv(DATA_DIR / 'customer_segments.csv')
df.head()
```

```python
features = ['annual_spend_10k', 'visits_per_month', 'online_ratio']

# 표준화 스케일링이 필요한 이유 : 거리 기반 알고리즘(k-means)이므로 변수 간 스케일(단위)
X_scaled = StandardScaler().fit_transform(df[features])
X_scaled
```

```python
results = []
for k in range (2,6):   # k=2, 3, 4, 5
    # n_clusters : 몇 개의 그룹(군집)으로 나눌지 정한다 .2, 3, 4, 5 --> 우리가 지정해야 한다.
    # n_init : 서로 다른 초기 중심점으로 10번 반복 실행 후 가장 좋은 결과를 선택(초기값에 따른 편차 완화)
    
    labels = KMeans(n_clusters=k, n_init=10, random_state=42).fit_predict(X_scaled)
    results.append({
        'k': k,
        'silhouette': silhouette_samples(X_scaled, labels)
    })  # K값에 따른 실루엣 점수를 비교해서 군집을 잘 나뉘는 K를 찾는 것이 목표

results
```

```python
pd.DataFrame(results)
```

### K=3으로 최종 군집화하고, 각 군집의 특징을 평균값으로 요약 (프로파일링)

```python
model = KMeans(n_clusters=3, n_init=10, random_state=42)
df.head()
```

```python
df['cluster'] = model.fit_predict(X_scaled) # 원본 df에 군집 번호 열 추가
df.head(10)
```

```python
profile = df.groupby('cluster')[features].mean().round(2)
```

```python
# 각 군집에 몇 명이 속하는지 표본 수 확인
profile['count'] = df['cluster'].value_counts().sort_index()
profile
```

```python
# 각 군집(cluster)의 평균 프로필을 보고 행동 중심으로 이름 붙이기 예시
names = {
    0: '온라인 중심 저빈도 고객 : 연간 고매액과 방문 횟수는 낮지만 온라인 구매 비율이 높다.',
    1: '매장 중심 우수 고객 : 연간 구매액과 방문 횟수가 가장 높고 온리안 비율은 낮다.',
    2: '균형형 일반 고객 : 고매액과 방문 횟수, 온라인 비율이 모두 중간 수준이다'
}
df['cluster_name'] = df['cluster'].map(names)   # 숫자 군집 번호를 사람이 읽을 수 있는 이름으로 매핑
df.head(10)
```
